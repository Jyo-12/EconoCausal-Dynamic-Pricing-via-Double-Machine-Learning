```python

import pandas as pd
from pathlib import Path

pd.set_option("display.max_columns", 100)

DATA_DIR = Path("data")

if not DATA_DIR.exists():
    raise FileNotFoundError(
        f"Dataset directory not found: {DATA_DIR.resolve()}"
    )

file_names = {
    "transactions": "transaction_data.csv",
    "products": "product.csv",
    "promotions": "causal_data.csv",
    "demographics": "hh_demographic.csv",
    "campaign_desc": "campaign_desc.csv",
    "campaign_table": "campaign_table.csv",
    "coupons": "coupon.csv",
    "coupon_redemptions": "coupon_redempt.csv"
}

datasets = {}

for name, filename in file_names.items():
    path = DATA_DIR / filename

    if not path.exists():
        print(f"Missing file: {filename}")
        continue

    print(f"Loading {filename}...")

    try:
        df = pd.read_csv(
            path,
            chunksize=25000,
            low_memory=True
        )

        datasets[name] = pd.concat(df, ignore_index=True)

        print(
            f"{name}: {datasets[name].shape[0]:,} rows, "
            f"{datasets[name].shape[1]} columns"
        )

    except Exception as e:
        print(f"Failed to load {filename}: {e}")
```

```python
for name, df in datasets.items():
    print("\n" + "=" * 80)
    print(f"DATASET: {name.upper()}")
    print("=" * 80)

    print("\nShape:", df.shape)

    print("\nColumn names:")
    print(df.columns.tolist())

    print("\nInformation:")
    df.info()

    print("\nDescriptive statistics:")
    display(df.describe(include="all").T)

    print("\nMissing values:")
    missing = pd.DataFrame({
        "missing_count": df.isnull().sum(),
        "missing_percentage": (
            df.isnull().mean() * 100
        ).round(2)
    }).sort_values("missing_percentage", ascending=False)

    display(missing)

    print("\nFully duplicated rows:", df.duplicated().sum())
```

```python
coupons = datasets["coupons"].copy()

print("Original rows:", len(coupons))
print("Exact duplicate rows:", coupons.duplicated().sum())

coupons_clean = coupons.drop_duplicates().copy()

print("Cleaned rows:", len(coupons_clean))
print("Rows removed:", len(coupons) - len(coupons_clean))

print(
    "Remaining exact duplicates:",
    coupons_clean.duplicated().sum()
)
```

```python
# Check key uniqueness in the cleaned coupon dataset

print("Rows:", len(coupons_clean))

print(
    "Unique COUPON_UPC values:",
    coupons_clean["COUPON_UPC"].nunique()
)

print(
    "Duplicate COUPON_UPC rows:",
    coupons_clean.duplicated(subset=["COUPON_UPC"]).sum()
)

print(
    "Duplicate COUPON_UPC + PRODUCT_ID + CAMPAIGN combinations:",
    coupons_clean.duplicated(
        subset=["COUPON_UPC", "PRODUCT_ID", "CAMPAIGN"]
    ).sum()
)
```

```python
# Check uniqueness of important identifiers

key_checks = {
    "transactions": ["household_key", "BASKET_ID", "PRODUCT_ID"],
    "products": ["PRODUCT_ID"],
    "demographics": ["household_key"],
    "campaign_desc": ["CAMPAIGN"],
    "campaign_table": ["household_key", "CAMPAIGN"],
    "coupon_redemptions": [
        "household_key", "DAY", "COUPON_UPC", "CAMPAIGN"
    ],
}

for name, keys in key_checks.items():
    df = datasets[name]

    print(f"\n--- {name} ---")
    print("Rows:", len(df))

    for key in keys:
        print(
            f"Unique {key}: {df[key].nunique()}, "
            f"missing: {df[key].isna().sum()}"
        )

    print(
        f"Duplicate rows using {keys}:",
        df.duplicated(subset=keys).sum()
    )
```

```python
# Validate key relationships between datasets

transactions = datasets["transactions"]
products = datasets["products"]
demographics = datasets["demographics"]
campaign_desc = datasets["campaign_desc"]
campaign_table = datasets["campaign_table"]
coupons_clean = coupons_clean
coupon_redemptions = datasets["coupon_redemptions"]

def check_reference(source_df, source_col, reference_df, reference_col, label):
    source_values = source_df[source_col].dropna().unique()
    reference_values = set(reference_df[reference_col].dropna().unique())

    unmatched = [value for value in source_values if value not in reference_values]

    print(f"\n{label}")
    print("Unique source values:", len(source_values))
    print("Unmatched source values:", len(unmatched))
    print("Sample unmatched values:", unmatched[:10])


check_reference(
    transactions, "PRODUCT_ID",
    products, "PRODUCT_ID",
    "Transactions -> Products"
)

check_reference(
    transactions, "household_key",
    demographics, "household_key",
    "Transactions -> Demographics"
)

check_reference(
    campaign_table, "household_key",
    demographics, "household_key",
    "Campaign table -> Demographics"
)

check_reference(
    campaign_table, "CAMPAIGN",
    campaign_desc, "CAMPAIGN",
    "Campaign table -> Campaign descriptions"
)

check_reference(
    coupons_clean, "PRODUCT_ID",
    products, "PRODUCT_ID",
    "Coupons -> Products"
)

check_reference(
    coupon_redemptions, "COUPON_UPC",
    coupons_clean, "COUPON_UPC",
    "Coupon redemptions -> Coupons"
)

check_reference(
    coupon_redemptions, "CAMPAIGN",
    campaign_desc, "CAMPAIGN",
    "Coupon redemptions -> Campaign descriptions"
)
```

```python
# Compare household coverage without merging datasets

transaction_households = set(
    datasets["transactions"]["household_key"].unique()
)

demographic_households = set(
    datasets["demographics"]["household_key"].unique()
)

campaign_households = set(
    datasets["campaign_table"]["household_key"].unique()
)

print("Transaction households:", len(transaction_households))
print("Demographic households:", len(demographic_households))
print("Campaign households:", len(campaign_households))

print(
    "\nTransaction households missing demographics:",
    len(transaction_households - demographic_households)
)

print(
    "Campaign households missing demographics:",
    len(campaign_households - demographic_households)
)

print(
    "Households shared by transactions and campaign table:",
    len(transaction_households & campaign_households)
)

print(
    "Demographic households present in transactions:",
    len(demographic_households & transaction_households)
)

print(
    "Demographic households present in campaign table:",
    len(demographic_households & campaign_households)
)
```

```python
# Inspect transaction data types and unusual values

transactions = datasets["transactions"]

print("Column data types:")
print(transactions.dtypes)

print("\nNegative quantities:")
print((transactions["QUANTITY"] < 0).sum())

print("\nZero quantities:")
print((transactions["QUANTITY"] == 0).sum())

print("\nNegative sales values:")
print((transactions["SALES_VALUE"] < 0).sum())

print("\nZero sales values:")
print((transactions["SALES_VALUE"] == 0).sum())

print("\nNegative DAY values:")
print((transactions["DAY"] < 0).sum())

print("\nNegative WEEK_NO values:")
print((transactions["WEEK_NO"] < 0).sum())

print("\nNegative STORE_ID values:")
print((transactions["STORE_ID"] < 0).sum())
```

```python
# Investigate zero quantities and zero sales values

transactions = datasets["transactions"]

zero_quantity = transactions["QUANTITY"] == 0
zero_sales = transactions["SALES_VALUE"] == 0

print("Zero quantity AND zero sales:",
      (zero_quantity & zero_sales).sum())

print("Zero quantity BUT nonzero sales:",
      (zero_quantity & ~zero_sales).sum())

print("Nonzero quantity BUT zero sales:",
      (~zero_quantity & zero_sales).sum())

print("\nExamples with zero quantity:")
display(
    transactions.loc[zero_quantity].head(10)
)

print("\nExamples with zero sales:")
display(
    transactions.loc[zero_sales].head(10)
)
```

```python
# Verify missing values and inspect zero-quantity records with nonzero sales

transactions = datasets["transactions"]

print("Missing values by column:")
print(transactions.isna().sum())

zero_qty_nonzero_sales = transactions.loc[
    (transactions["QUANTITY"] == 0)
    & (transactions["SALES_VALUE"] != 0)
]

print("\nZero quantity but nonzero sales:", len(zero_qty_nonzero_sales))

display(zero_qty_nonzero_sales.head(15))
```

```python
# Inspect near-zero sales values among zero-quantity records

zero_qty_nonzero_sales = transactions.loc[
    (transactions["QUANTITY"] == 0)
    & (transactions["SALES_VALUE"] != 0)
]

near_zero = zero_qty_nonzero_sales["SALES_VALUE"].abs() < 1e-10

print("Zero quantity, nonzero sales:", len(zero_qty_nonzero_sales))
print("Near-zero sales values:", near_zero.sum())
print("Meaningful nonzero sales values:", (~near_zero).sum())

print("\nSmallest absolute sales values:")
display(
    zero_qty_nonzero_sales.assign(
        abs_sales=zero_qty_nonzero_sales["SALES_VALUE"].abs()
    ).sort_values("abs_sales").head(10)
)
```

```python
# Check product data types and empty product-size values

products = datasets["products"]

print("Data types:")
print(products.dtypes)

print("\nMissing values:")
print(products.isna().sum())

size = products["CURR_SIZE_OF_PRODUCT"]

print("\nEmpty strings:", size.eq("").sum())
print("Whitespace-only strings:", size.str.fullmatch(r"\s+").fillna(False).sum())
print("Unique product IDs:", products["PRODUCT_ID"].nunique())
print("Duplicate product IDs:", products["PRODUCT_ID"].duplicated().sum())

print("\nExamples of empty product sizes:")
display(products.loc[size.eq("")].head(10))
```

```python
import numpy as np

products_clean = datasets["products"].copy()

# Convert blank and whitespace-only product sizes to missing values
products_clean["CURR_SIZE_OF_PRODUCT"] = (
    products_clean["CURR_SIZE_OF_PRODUCT"]
    .replace(r"^\s*$", np.nan, regex=True)
)

print(
    "Product sizes normalized to missing:",
    products_clean["CURR_SIZE_OF_PRODUCT"].isna().sum()
)

print(
    "Remaining whitespace-only values:",
    products_clean["CURR_SIZE_OF_PRODUCT"]
    .str.fullmatch(r"\s+")
    .fillna(False)
    .sum()
)

print("Rows retained:", len(products_clean))
print(
    "Duplicate product IDs:",
    products_clean["PRODUCT_ID"].duplicated().sum()
)
```

```python
# Validate data types and missing values in remaining datasets

datasets_to_check = {
    "demographics": datasets["demographics"],
    "campaign_desc": datasets["campaign_desc"],
    "campaign_table": datasets["campaign_table"],
    "coupon_redemptions": datasets["coupon_redemptions"],
    "coupons_clean": coupons_clean,
}

for name, df in datasets_to_check.items():
    print(f"\n{'=' * 50}")
    print(f"Dataset: {name}")
    print("Shape:", df.shape)

    print("\nData types:")
    print(df.dtypes)

    print("\nMissing values:")
    print(df.isna().sum())

    print("\nExact duplicate rows:", df.duplicated().sum())
```

```python
# Validate the large promotion dataset in place
# No merge and no full DataFrame copy

promotions = datasets["promotions"]

print("Shape:", promotions.shape)

print("\nData types:")
print(promotions.dtypes)

print("\nMissing values:")
print(promotions.isna().sum())

print("\nUnique values in display:")
print(promotions["display"].value_counts(dropna=False))

print("\nUnique values in mailer:")
print(promotions["mailer"].value_counts(dropna=False))

print("\nUnique product IDs:", promotions["PRODUCT_ID"].nunique())
print("Unique store IDs:", promotions["STORE_ID"].nunique())
print("Unique week numbers:", promotions["WEEK_NO"].nunique())

print("\nNegative product IDs:",
      (promotions["PRODUCT_ID"] < 0).sum())

print("Negative store IDs:",
      (promotions["STORE_ID"] < 0).sum())

print("Negative week numbers:",
      (promotions["WEEK_NO"] < 0).sum())
```

```python
# Check uniqueness of the candidate promotion key

promotion_key = ["PRODUCT_ID", "STORE_ID", "WEEK_NO"]

print(
    "Duplicate product-store-week combinations:",
    promotions.duplicated(subset=promotion_key).sum()
)
```

```python
# Investigate duplicate promotion keys without copying the full dataset

promotion_key = ["PRODUCT_ID", "STORE_ID", "WEEK_NO"]

duplicate_mask = promotions.duplicated(
    subset=promotion_key,
    keep=False
)

print("Rows belonging to repeated keys:", duplicate_mask.sum())

print("\nExamples of repeated-key records:")
display(promotions.loc[duplicate_mask].sort_values(promotion_key).head(20))
```

```python
# Count repeated keys that have different display/mailer combinations

duplicate_rows = promotions.loc[duplicate_mask]

distinct_combinations = duplicate_rows.drop_duplicates(
    subset=promotion_key + ["display", "mailer"]
)

print("Distinct product-store-week-promotion combinations:",len(distinct_combinations))

print("Repeated-key rows with the same promotion values may be exact duplicates.")
```

```python
# Check exact duplicate rows across all five promotion columns

print(
    "Exact duplicate promotion rows:",
    promotions.duplicated().sum()
)

# Check whether the full five-column combination is unique
print(
    "Duplicate full promotion combinations:",
    promotions.duplicated(
        subset=[
            "PRODUCT_ID",
            "STORE_ID",
            "WEEK_NO",
            "display",
            "mailer"
        ]
    ).sum()
)
```

```python
# Validate campaign start and end days

campaign_desc = datasets["campaign_desc"]

print("Campaign rows:", len(campaign_desc))
print("Unique campaign IDs:", campaign_desc["CAMPAIGN"].nunique())

print(
    "Campaigns with START_DAY > END_DAY:",
    (campaign_desc["START_DAY"] > campaign_desc["END_DAY"]).sum()
)

print(
    "Campaigns with START_DAY <= 0:",
    (campaign_desc["START_DAY"] <= 0).sum()
)

print(
    "Campaigns with END_DAY <= 0:",
    (campaign_desc["END_DAY"] <= 0).sum()
)

print("\nCampaign day ranges:")
display(
    campaign_desc[
        ["CAMPAIGN", "START_DAY", "END_DAY"]
    ].sort_values("CAMPAIGN")
)
```

```python
# Validate transaction day, week, and time ranges

transactions = datasets["transactions"]

for col in ["DAY", "WEEK_NO", "TRANS_TIME"]:
    print(f"\n--- {col} ---")
    print("Minimum:", transactions[col].min())
    print("Maximum:", transactions[col].max())
    print("Unique values:", transactions[col].nunique())

print("\nTRANS_TIME below 0:",
      (transactions["TRANS_TIME"] < 0).sum())

print("TRANS_TIME above 2359:",
      (transactions["TRANS_TIME"] > 2359).sum())

print("TRANS_TIME with invalid minute component:",
      ((transactions["TRANS_TIME"] % 100) >= 60).sum())
```

```python
# Validate key numeric fields and campaign references

transactions = datasets["transactions"]
campaign_table = datasets["campaign_table"]
coupon_redemptions = datasets["coupon_redemptions"]
campaign_desc = datasets["campaign_desc"]

print("TRANSACTION VALUE CHECKS")
for col in [
    "QUANTITY", "SALES_VALUE", "RETAIL_DISC",
    "COUPON_DISC", "COUPON_MATCH_DISC"
]:
    print(
        col,
        "| min:", transactions[col].min(),
        "| max:", transactions[col].max()
    )

print("\nCAMPAIGN TABLE")
print(
    "Campaign IDs missing from descriptions:",
    len(
        set(campaign_table["CAMPAIGN"])
        - set(campaign_desc["CAMPAIGN"])
    )
)
print(
    "Campaigns with START_DAY > END_DAY:",
    (campaign_desc["START_DAY"] > campaign_desc["END_DAY"]).sum()
)

print("\nCOUPON REDEMPTIONS")
print(
    "Redemption days outside observed transaction DAY range:",
    (
        (coupon_redemptions["DAY"] < transactions["DAY"].min())
        | (coupon_redemptions["DAY"] > transactions["DAY"].max())
    ).sum()
)
```

```python
from pathlib import Path
import json

output_dir = Path("cleaned_data")
output_dir.mkdir(exist_ok=True)

# Save datasets separately; do not merge them
datasets_to_save = {
    "transactions": datasets["transactions"],
    "products": products_clean,
    "promotions": datasets["promotions"],
    "demographics": datasets["demographics"],
    "campaign_desc": datasets["campaign_desc"],
    "campaign_table": datasets["campaign_table"],
    "coupons": coupons_clean,
    "coupon_redemptions": datasets["coupon_redemptions"],
}

for name, df in datasets_to_save.items():
    path = output_dir / f"{name}_clean.csv"
    df.to_csv(path, index=False)
    print(f"Saved: {path} | Rows: {len(df):,}")

print("\nAll eight datasets saved separately.")
```
